```python
import pandas as pd
```

```python

df = pd.read_csv('adidas_dataset_1200rows.csv')
df.head()
```

```python
total_revenue = df['Revenue'].sum()
```

```python
total_profit = df['Profit'].sum()
```

```python
total_units = df['Units_Sold'].sum()
```

```python
avg_order_value = df['Revenue'].mean()
```

```python
print("Performance of sales channel by store type")
store_summary = df.groupby('Store_Type')[['Revenue', 'Profit', 'Units_Sold']].sum().sort_values(by='Revenue', ascending=False)
print(store_summary)
```

```python
print("Performance by Region ")
region_summary = df.groupby('Region')[['Revenue', 'Profit', 'Units_Sold']].sum().sort_values(by='Revenue', ascending=False)
print(region_summary)
print("\n")
```

```python
print(" Top 5 Best-Selling Products ")
top_products = df.groupby('Product_Name')[['Revenue', 'Profit', 'Units_Sold']].sum().sort_values(by='Revenue', ascending=False).head(5)
print(top_products)
print("\n")
```

```python
print(" Performance by Category ")
category_summary = df.groupby('Category')[['Revenue', 'Profit', 'Units_Sold']].sum().sort_values(by='Revenue', ascending=False)
print(category_summary)
print("\n")
```

### Gross Revenue


```python
df['Gross_Revenue'] = df['Units_Sold'] * df['Unit_Price']
```

### Total Discount Amount 


```python
df['Discount_Amount'] = (df['Gross_Revenue'] * (df['Discount'] / 100)).round(2)
```

### Profit Margin


```python
df['Profit_Margin_%'] = ((df['Profit'] / df['Revenue']) * 100).round(2)
```
